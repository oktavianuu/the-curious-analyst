## Pengujuan Hipotesis
#### Prinsip Dasar Hipotesis
Dalam statistik, hipotesis adalah pernyataan yang dapat diuji mengenai parameter populasi. Proses pengujian ini selalu melibatkan sepasang hipotesis yang saling terpisah:
- **Hipotesis Nol ($H_0$)**: Ini adalah asumsi dasar atau status quo. $H_0$ selalu menyatakan bahwa tidak ada efek, tidak ada perbedaan, atau tidak ada hubungan akibat dari perlakuan eksperimental yang diberikan. Dalam analisis, kita selalu beroperasi di bawah asumsi bahwa $H_0$ adalah benar, sampai data membuktikan sebaliknya.
- **Hipotesis Alternatif ($H_1$)**: Ini adalah prediksi dari peneliti. $H_1$ menyatakan bahwa perlakuan eksperimental memiliki efek, atau terdapat perbedaan/hubungan yang nyata pada populasi.

#### Tingkat Signifikansi ($\alpha$) dan Toleransi Risiko
Sebelum mengumpulkan data, seorang peneliti wajib menetapkan ambang batas risiko yang disebut Tingkat Signifikansi ($\alpha$). Secara matematis, $\alpha$ mendefinisikan probabilitas terjadinya Kesalahan Tipe I (menolak $H_0$ padahal $H_0$ sebenarnya benar di populasi). Dengan kata lain, ini adalah probabilitas maksimal yang berani kita tanggung untuk mengklaim bahwa "ada efek", padahal kenyataannya perlakuan tersebut tidak menghasilkan efek apa-apa. Nilai standar yang paling sering digunakan di berbagai disiplin ilmu adalah $\alpha = 0.05$ (toleransi risiko false positive sebesar 5%).

Nilai $\alpha$ tidaklah kaku, melainkan harus disesuaikan dengan cost (konsekuensi) dari Kesalahan Tipe I:
- **Lebih Longgar ($\alpha = 0.10$ atau $0.20$)**: Sering digunakan dalam studi pendahuluan (pilot studies). Di sini, peneliti lebih mementingkan eksplorasi dan mencari potensi efek, sehingga risiko Kesalahan Tipe I tidak terlalu berakibat fatal.
- **Sangat Ketat ($\alpha = 0.01$ atau $0.001$)**: Diwajibkan dalam pengujian medis atau obat-obatan. Mengklaim sebuah obat efektif padahal aslinya tidak (Kesalahan Tipe I) dapat membahayakan nyawa pasien. Oleh karena itu, standar pembuktiannya harus dibuat sekuat mungkin.

#### P-Value
- P-value adalah besarnya probabilitas hasil penelitian kita muncul murni karena faktor kebetulan, dengan asumsi bahwa sebenarnya tidak ada efek atau perbedaan nyata di dalam populasi ($H_0$).
- Formulasi matematisnya adalah: $P(\text{Data} \mid H_0)$, bukan $P(H_0 \mid \text{Data})$. 
- P-value mengukur seberapa "kaget" kita melihat data yang ada, jika kita asumsikan tidak ada efek sama sekali. Semakin kecil p-value, semakin data tersebut tidak selaras dengan asumsi $H_0$.
- Untuk menentukan apakah hasil pengujian bersifat "signifikan secara statistik", kita membandingkan p-value empiris (hasil perhitungan data) dengan $\alpha$ teoretis (ambang batas risiko).
- Jika $p\text{-value} \le \alpha$: Data yang diamati sangat tidak masuk akal jika $H_0$ benar. Oleh karena itu, kita Menolak $H_0$. Kita menyimpulkan bahwa terdapat cukup bukti empiris untuk mendukung $H_1$.
- Jika $p\text{-value} > \alpha$: Data yang diamati masih sangat mungkin terjadi hanya karena variasi acak (kebetulan) di bawah asumsi $H_0$. Oleh karena itu, kita Gagal Menolak $H_0$.

**NOTES**:
Dalam kerangka berpikir statistik frequentist, ketiadaan bukti adanya efek (gagal menolak $H_0$) bukanlah bukti bahwa efek tersebut tidak ada. Kita tidak pernah bisa membuktikan $H_0$ benar, kita hanya kekurangan bukti data untuk membantahnya.

## Dua Jenis Kesalahan dalam Pengujian Hipotesis
Dalam statistik inferensial, keputusan kita didasarkan pada sampel dan probabilitas, sehingga tidak ada kepastian absolut. Oleh karena itu, kita selalu dihadapkan pada risiko membuat kesalahan saat mengambil kesimpulan mengenai Hipotesis Nol ($H_0$) dan Hipotesis Alternatif ($H_1$). Berikut adalah dua jenis kesalahan utama dalam pengujian hipotesis:

**1) Kesalahan Tipe I (Type I Error / False Positive)**
- Kesalahan ini terjadi ketika kita menolak $H_0$ padahal $H_0$ sebenarnya benar di populasi. Dengan kata lain, kita secara keliru menyimpulkan bahwa ada efek, perbedaan, atau hubungan yang signifikan, padahal kenyataannya hal tersebut tidak ada. Hal ini dikenal dengan  alarm palsu (false positive). 
- Peluang terjadinya Kesalahan Tipe I ditetapkan oleh peneliti dan disimbolkan dengan $\alpha$ (Alpha), yang juga dikenal sebagai tingkat signifikansi (significance level).
- Secara statistik, kesalahan ini bisa murni terjadi karena kebetulan (sampel acak yang secara tidak sengaja menunjukkan hasil ekstrem). Namun, probabilitas munculnya temuan false positive dalam penelitian aktual sering kali diperparah oleh kelemahan desain, seperti teknik pengambilan sampel yang bias sehingga tidak merepresentasikan populasi, atau prosedur implementasi penelitian yang kurang ketat.

**2) Kesalahan Tipe II (Type II Error / False Negative)**
- Kesalahan ini terjadi ketika kita gagal menolak $H_0$ padahal $H_0$ sebenarnya salah (atau secara implisit keliru "menerima" $H_0$). Ini berarti penelitian kita gagal mendeteksi efek atau hubungan yang sebenarnya benar-benar ada di populasi. Hal ini dikenal dengan kegagalan deteksi (false negative). 
- Peluang terjadinya Kesalahan Tipe II disimbolkan dengan $\beta$ (Beta).
- Risiko ini sangat dipengaruhi oleh ukuran sampel (sample size) dan besaran efek (effect size). Kesalahan Tipe II paling sering terjadi karena ukuran sampel yang terlalu kecil untuk mendeteksi effect size yang rendah atau moderat.

Kedua masalah di atas (sampel kecil dan efek rendah) berdampak langsung pada penurunan Daya Uji Statistik (Statistical Power). Power adalah probabilitas bahwa penelitian kita akan sukses mendeteksi efek yang benar-benar ada (dirumuskan sebagai $1 - \beta$). Ketika Power rendah, risiko Kesalahan Tipe II ($\beta$) meningkat secara drastis.

#### NOTES: 
Terdapat hubungan trade-off (tarik-ulur) antara kedua kesalahan ini. Jika kita berusaha terlalu keras meminimalkan risiko Kesalahan Tipe I dengan memperketat nilai $\alpha$ (misalnya menjadikannya 0.01 dari 0.05), kita secara otomatis membuat $H_0$ lebih sulit ditolak, yang berarti kita meningkatkan risiko Kesalahan Tipe II ($\beta$). Untuk menghindari hal  ini, kita harus secara cermat  merancang eksperimen yang menyeimbangkan keduanya melalui perhitungan sampel yang tepat sejak awal. 

##  Statistical Power, atau Daya Uji ($1 - \beta$)
Secara statistik, Power didefinisikan sebagai probabilitas untuk menolak Hipotesis Nol ($H_0$) secara tepat, dengan asumsi bahwa Hipotesis Alternatif ($H_1$) adalah benar. Power adalah kepekaan eksperimen kita untuk menemukan kebenaran. Nilai Power dihitung dengan rumus $1 - \beta$, di mana $\beta$ (Beta) adalah probabilitas terjadinya Kesalahan Tipe II (kegagalan mendeteksi efek yang nyata).

Jika penelitian kita memiliki Power sebesar $0.80$ (atau $80\%$), ini berarti apabila perlakuan atau intervensi yang kita teliti benar-benar memiliki efek, kita memiliki peluang sebesar $80\%$ untuk mendapatkan hasil yang signifikan secara statistik, dan risiko sebesar $20\%$ untuk gagal melihat efek tersebut.

Dalam literatur ilmiah modern, terdapat sebuah konvensi tidak tertulis namun sangat direkomendasikan untuk digunakan yakni bahwa sebuah penelitian yang baik harus dirancang sejak awal untuk memiliki minimal Power sebesar $80\%$ ($0.80$).Melakukan eksperimen tanpa menghitung Power adalah sebuah kecerobohan metodologis. Jika kita menjalankan penelitian dengan Power hanya $30\%$, kita sebenarnya sedang melakukan penelitian yang dirancang untuk gagal. Kita menghabiskan dana, waktu, dan melibatkan subjek (seperti pasien atau hewan uji) untuk sebuah instrumen yang sejak awal sebagian besar sudah "buta" terhadap efek yang ingin kita buktikan. Oleh karena itu, sebelum pengumpulan data dimulai, kita wajib melakukan analisis Power (Power Analysis) untuk menjawab satu pertanyaan logistik yang paling krusial: "Mengingat efek yang kita perkirakan sebesar X, dan batas signifikansi yang kita inginkan sebesar Y, berapa ukuran sampel ($N$) minimal yang harus kita kumpulkan agar uji kita tidak meleset?"

## Penentuan Ukuran Sampel dalam Penelitian Survey
$$N = \frac{Z_{\alpha/2}^2 \times P \times (1 - p) \times D}{E^2}$$

Desain penelitian sangat bergantung pada nilai $D$ (Design Effect).
**1) Desain Simple Random Sampling (SRS) $\rightarrow$ $D = 1$**

Jika kita menetapkan $D = 1$, rumus menuntut kita untuk memiliki daftar seluruh populasi. Konsekuensinya adalah kita harus meminta data nama siswa dari Dinas Pendidikan, lalu mengundi nama-nama tersebut satu per satu dengan komputer. Keuntungan metode ini adalah bahwa jumlah sampel yang dibutuhkan ($N$) adalah yang paling minimal. Namun kerugiannya adalah bahwa siswa  yang terpilih akan tersebar secara acak. Kita harus mengunjungi 30 sekolah berbeda hanya untuk menemui 2 atau 3 siswa terpilih di masing-masing sekolah.

**2) Desain Cluster Sampling $\rightarrow$ $D = 2$ (atau lebih)**

Banyak peneliti pendidikan menggunakan metode ini. Alih-alih mengacak nama siswa, kita mengacak nama sekolah atau nama kelas. Misalnya  memilih 5 sekolah secara acak. Ketika kita datang ke sekolah tersebut, kita membagikan angket miskonsepsi ke seluruh siswa di kelas yang bersangkutan. Hal ini mempermudah peneliti secara logistik. Kita hanya perlu mengurus izin ke 5 kepala sekolah, bukan 30. Namun kita harus menyertakan nilai $D$ (misalnya $D=2$). Ini berarti target jumlah siswa kita akan berlipat ganda. Mengapa dihukum dua kali lipat? Karena siswa dalam satu sekolah yang sama (diajar oleh guru yang sama, kurikulum sama) cenderung memiliki pola pikir dan miskonsepsi yang mirip. Secara statistik, 30 siswa dari kelas yang sama tidak memberikan informasi yang sekaya 30 siswa dari 30 sekolah yang berbeda.

## Simulasi 
Penelitian: "Miskonsepsi Siswa SMA Terhadap Materi Evolusi di Kota Pontianak.
#### Langkah 1: Menetapkan Parameter Dasar 
- $Z_{\alpha/2}$: kita ingin tingkat kepercayaan $95\%$. Secara statistik baku, nilai $Z = 1.96$.
- $P$ (Prevalensi): kita belum tahu berapa persen siswa yang punya miskonsepsi ini. Untuk mencari target sampel terbesar (paling aman), kita tetapkan $P = 0.5$ ($50\%$).
- $E$ (Presisi/Margin of Error): Kita menoleransi kesalahan sebesar $5\%$ saja. $E = 0.05$.

#### Langkah 2: Menghitung Sampel Ideal (SRS murni / $D=1$)
$$N = \frac{1.96^2 \times 0.5 \times (1 - 0.5) \times 1}{0.05^2}$$
$$N = \frac{3.8416 \times 0.25}{0.0025}$$
$$N = \frac{0.9604}{0.0025} \approx 384 \text{ siswa}$$

#### Langkah 3: Menetapkan $D$ 
Kita menyadari tidak mungkin mencari 384 siswa yang terpencar satu per satu di puluhan SMA di Pontianak. Kita memutuskan menggunakan Cluster Sampling. Kita akan memilih beberapa SMA secara acak, lalu mengambil satu angkatan penuh di sekolah itu. Karena kita menggunakan cluster, literatur menyarankan kita menggunakan nilai Design Effect $D = 2$.

#### Langkah 4: Perhitungan Final
$$N = \text{Sampel Ideal} \times D$$
$$N = 384 \times 2 = 768 \text{ siswa}$$

#### Langkah 5: Strategi Pemilihan Siswa di Lapangan
Sekarang Kita tahu bahwa kita membutuhkan 768 siswa. Bagaimana mendapatkannya? 
- Kita mencari data rata-rata jumlah siswa per sekolah (misalnya satu angkatan kelas XI rata-rata berisi 150 siswa).
- Kita membagi total kebutuhan: $768 / 150 \approx 5.12$.
- Keputusan Desain: Kita harus mengundi dan memilih 6 SMA di kota Pontianak secara acak. Kita mendatangi keenam sekolah tersebut, masuk ke semua kelas XI, dan menyebarkan angket secara total. Kita akan mendapatkan sekitar 900 lembar angket (melebihi target 768, yang sangat bagus untuk berjaga-jaga jika ada siswa yang mengisi asal-asalan).
