```r
df_male <- read_excel("../data/male_final.xlsx")
head(df)
```

```r
# Recoding ketiga variabel
df_male <- df_male %>%
  mutate(across(c(citra_tubuh, aktivitas_fisik, gizi_seimbang),
                ~ case_when(
                  . %in% c(1, 2) ~ "Rendah",
                  . == 3        ~ "Sedang",
                  . %in% c(4, 5) ~ "Tinggi"
                ),
                .names = "{.col}_cat")) %>%  # membuat kolom baru dengan akhiran "_cat"
  mutate(across(ends_with("_cat"),
                ~ factor(., levels = c("Rendah", "Sedang", "Tinggi"))))

# menetapkan reference (baseline)
# Membandingkan semua kategori terhadap 'rendah'
df_male$citra_tubuh_cat <- relevel(df_male$citra_tubuh_cat, ref = "Rendah")
```

```r
head(df_male)
```

```r
write_xlsx(df_male, "../data/data_male_final.xlsx")
```

```r
# 3. Memfit model
model_male <- multinom(citra_tubuh_cat ~ aktivitas_fisik_cat + gizi_seimbang_cat, data = df_male)
```

```r
# 4. Ringkasan Model
summary(model_male)
```

```r
# (Logits, SE, p-values)
tidy_table <- tidy(model_male, conf.int = TRUE)

# Menambahkan Odds Ratios and Confidence Intervals
# Math: OR = exp(estimate)
tidy_table <- tidy_table %>%
  mutate(
    odds_ratio = exp(estimate),
    OR_conf_low = exp(conf.low),
    OR_conf_high = exp(conf.high)
  ) %>%
  select(y.level, term, estimate, std.error, p.value, odds_ratio, OR_conf_low, OR_conf_high)

# Hasil
print(tidy_table)
```

```r
View(tidy_table)
```

```r
write_xlsx(tidy_table, "tabel_regresi_male.xlsx")
```

```r
# GANTI 'df_anda' dengan nama variabel data frame milikmu (misal: df atau data_skripsi)
df_konten <- df_male

# --- 1. Tabel Deskriptif (Karakteristik Responden) ---
tab_deskriptif <- as.data.frame(table(df_konten$gizi_seimbang)) %>%
  rename(Kategori = Var1, Frekuensi = Freq) %>%
  mutate(Persentase = round((Frekuensi / sum(Frekuensi)) * 100, 2))

# --- 2. Tabel Tabulasi Silang (Gizi vs Citra Tubuh) ---
tab_cross <- df_konten %>%
  group_by(gizi_seimbang, citra_tubuh) %>%
  summarise(n = n(), .groups = 'drop') %>%
  pivot_wider(names_from = citra_tubuh, values_from = n, values_fill = 0)

# --- 3. Tabel Kelayakan Model (AIC/BIC) ---
tab_fit <- data.frame(
  Indikator = c("AIC", "BIC", "Log-Likelihood", "Deviance"),
  Nilai = round(c(AIC(model), BIC(model), as.numeric(logLik(model)), deviance(model)), 3)
)

# --- EKSPOR SEMUA KE SATU FILE EXCEL (Beda Sheet) ---
list_hasil_akhir <- list(
  "Regresi_Utama" = tidy_table, # Tabel Odds Ratio yang sudah kita buat
  "Deskriptif" = tab_deskriptif,
  "Tabulasi_Silang" = tab_cross,
  "Uji_Kelayakan_Model" = tab_fit
)

write_xlsx(list_hasil_akhir, "Hasil_Lengkap_Penelitian.xlsx")
```

```r
# Membuat Forest Plot Estetik
p1 <- plot_model(model_male, 
           type = "est",         
           transform = "exp",    
           show.values = TRUE,   
           value.offset = 0.4,    # Diperlebar sedikit agar angka tidak menabrak titik
           show.p = TRUE,         # Menambahkan tanda bintang (*) untuk signifikansi
           dot.size = 2.5,        # Memperbesar ukuran titik
           line.size = 1,         # Mempertebal garis Confidence Interval
           colors = "#2A9D8F",    # Menggunakan warna Hijau Tosca elegan untuk titik & garis
           vline.color = "#E63946", # Merah tegas untuk garis batas (Null Line)
           title = "Pengaruh Gaya Hidup terhadap Citra Tubuh Laki-laki",
           axis.title = "Odds Ratio (Skala Logaritmik)") +
  
  # Kosmetik kustomisasi tema
  theme_minimal(base_size = 14) +
  theme(
    # Mengatur Judul
    plot.title = element_text(face = "bold", hjust = 0.5, size = 16, color = "#2B2D42"),
    
    # Mengatur Sumbu (Axis)
    axis.title.x = element_text(face = "bold", margin = margin(t = 15)),
    axis.text.y = element_text(face = "bold", size = 12, color = "#2B2D42"),
    
    # Mengatur Judul Panel/Facet (Untuk membedakan target "Sedang" dan "Tinggi")
    strip.text = element_text(face = "bold", size = 13, color = "white"),
    strip.background = element_rect(fill = "#8D99AE", color = NA),
    
    # Membersihkan Latar Belakang (White Space)
    panel.grid.minor = element_blank(),
    panel.grid.major.x = element_blank(), # Menghilangkan garis vertikal (kecuali garis merah)
    panel.grid.major.y = element_line(linetype = "dotted", color = "gray80") # Garis bantu horizontal tipis
  )

# Menampilkan plot
print(p1)

# Simpan dengan resolusi tinggi
ggsave("../report/visualization/forest_plot_estetik.png", plot = p1, width = 10, height = 7, dpi = 300)
```

```r
# 1. Hitung bobot terbalik (Inverse Frequency)
tabel_bobot <- 1 / table(df_clean$citra_tubuh)

# 2. Bikin kolom baru di data frame yang isinya murni ANGKA (Ini kunci mengatasi errornya!)
df_clean$bobot_kategori <- as.numeric(tabel_bobot[df_clean$citra_tubuh])

# 3. Masukkan ke model menggunakan nama kolom bobot yang baru dibuat
model_weighted <- multinom(citra_tubuh ~ gizi_seimbang + aktivitas_fisik, 
                           data = df_clean, 
                           weights = bobot_kategori)
```

weights: 18 (10 variable): Algoritma membaca ada 10 parameter (koefisien) yang harus dia hitung untuk memecahkan persamaanmu.

initial value 3.29... turun menjadi final value 2.83...: Algoritma ini ibarat orang buta yang sedang mencari titik terendah di sebuah lembah. Nilai yang terus turun ini (disebut Deviance) membuktikan bahwa modelmu sedang "belajar" dan berhasil meminimalkan tingkat kesalahannya dari iterasi ke iterasi.

converged: Model sudah sampai di dasar lembah. Dia sudah menemukan kombinasi angka yang paling optimal dan berhenti mencari.

```r
# 1. Ambil prediksi dari model yang BARU (yang sudah diberi bobot)
prediksi_weighted <- predict(model_weighted)

# 2. Buat tabel konfusi baru
tab_weighted <- table(Kenyataan = df_clean$citra_tubuh, Prediksi = prediksi_weighted)
print("=== TABEL KONFUSI (MODEL DENGAN BOBOT) ===")
print(tab_weighted)

# 3. Hitung akurasi total yang baru
akurasi_weighted <- sum(diag(tab_weighted)) / sum(tab_weighted)
cat("\nAkurasi Total Baru:", round(akurasi_weighted * 100, 2), "%\n")

# 4. (Opsional tapi keren) Hitung akurasi KHUSUS untuk kategori Rendah
# Ganti angka 1 dengan urutan baris "Rendah" di tabelmu (biasanya baris ke-1)
akurasi_rendah <- tab_weighted[1, 1] / sum(tab_weighted[1, ])
cat("Akurasi Khusus Kategori Rendah:", round(akurasi_rendah * 100, 2), "%\n")
```
