```python
import pandas as pd
import numpy as np
```

```python
male =  pd.read_excel("../data/raw/male_data.xlsx")
female = pd.read_excel("../data//raw/female_data.xlsx")
```

```python
male.head()
```

```python
# Tidak perlu membuat slice terpisah
cols = ["citra_tubuh", "aktivitas_fisik", "gizi_seimbang"]
mapping = {1: "Rendah", 2: "Rendah", 3: "Sedang", 4: "Tinggi", 5: "Tinggi"}

for col in cols:
    male[f'{col}_cat'] = male[col].map(mapping).astype('object')
    male[f'{col}_cat'] = pd.Categorical(
        male[f'{col}_cat'],
        categories=["Rendah", "Sedang", "Tinggi"],
        ordered=True
    )

male
```

```python
male_df = male[['citra_tubuh', 'aktivitas_fisik', 'gizi_seimbang', 'citra_tubuh_cat', 'aktivitas_fisik_cat', 'gizi_seimbang_cat']]
male_df.to_excel("../data/processed/male.xlsx", index=False)
```

```python
# Tidak perlu membuat slice terpisah
cols = ["citra_tubuh", "aktivitas_fisik", "gizi_seimbang"]
mapping = {1: "Rendah", 2: "Rendah", 3: "Sedang", 4: "Tinggi", 5: "Tinggi"}

for col in cols:
    female[f'{col}_cat'] = female[col].map(mapping).astype('object')
    female[f'{col}_cat'] = pd.Categorical(
        female[f'{col}_cat'],
        categories=["Rendah", "Sedang", "Tinggi"],
        ordered=True
    )

female
```

```python
female_df = female[['citra_tubuh', 'aktivitas_fisik', 'gizi_seimbang', 'citra_tubuh_cat', 'aktivitas_fisik_cat', 'gizi_seimbang_cat']]
female_df.to_excel("../data/processed/female.xlsx", index=False)
```

```python
male.to_excel('../data/male_final.xlsx', index=False)
female.to_excel('../data/female_final.xlsx', index=False)
```
