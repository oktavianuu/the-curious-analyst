## FEMALE

```r
df <- read_excel('female_final.xlsx')
```

```r
model <- glm(pengalaman_biner ~ dukungan_biner + sumber_biner,
             data = df,
             family = 'binomial')

summary(model)
```

```r
# Menghitung Odds Ratio
OR <- exp(coef(model))

# Menghitung Confidence Interval (95%)
CI <- exp(confint(model))

# Menggabungkan hasil agar mudah dibaca
tabel_hasil <- cbind(OR, CI)
colnames(tabel_hasil) <- c("Odds Ratio", "Lower CI (2.5%)", "Upper CI (97.5%)")

print("--- Hasil Analisis Odds Ratio ---")
print(tabel_hasil)
```

```r
# 2. Ekstraksi hasil matematis model menjadi tabel yang rapi (Tidy Data)
# Argumen 'exponentiate = TRUE' otomatis mengubah log-odds menjadi Odds Ratio
hasil_model <- tidy(model, exponentiate = TRUE, conf.int = TRUE)

# 3. Manipulasi Data untuk Visualisasi
data_plot <- hasil_model %>%
  # Kita membuang Intercept karena tidak diinterpretasikan dalam Forest Plot inferensi
  filter(term != "(Intercept)") %>%
  mutate(
    # Merapikan nama variabel agar elegan di sumbu Y
    Variabel = case_when(
      term == "dukungan_biner" ~ "Dukungan Sosial\n(Baik vs Kurang)",
      term == "sumber_biner" ~ "Sumber Informasi\n(Baik vs Kurang)",
      TRUE ~ term
    ),
    # Membuat label teks persis di atas garis (OR [Lower, Upper])
    Label_Teks = sprintf("%.2f [%.2f, %.2f]", estimate, conf.low, conf.high),
    # Mengkategorikan warna berdasarkan P-Value (Batas 0.05)
    Signifikan = ifelse(p.value < 0.05, "Signifikan (p < 0.05)", "Tidak Signifikan")
  )

# 4. Merakit Grafik Lapis demi Lapis (The Grammar of Graphics)
plot_forest_female <- ggplot(data_plot, aes(x = estimate, y = Variabel, color = Signifikan)) +
  
  # A. Menambahkan garis vertikal merah di angka 1 (Garis Nol Efek)
  geom_vline(xintercept = 1, linetype = "dashed", color = "#c0392b", size = 1) +
  
  # B. Menggambar rentang Confidence Interval dan titik Odds Ratio
  geom_pointrange(aes(xmin = conf.low, xmax = conf.high), linewidth = 1.2, fatten = 4) +
  
  # C. Menambahkan Teks Angka agar pembaca tidak perlu menebak nilai pastinya
  geom_text(aes(label = Label_Teks), vjust = -1.8, size = 4, color = "black", fontface = "bold") +
  
  # D. KUNCI MATEMATIS: Mengubah sumbu X menjadi skala Logaritmik
  scale_x_log10(breaks = c(0.5, 1, 2, 5, 10)) +
  
  # E. Skema warna elegan (Biru Tua untuk Signifikan, Abu-abu untuk Tidak)
  scale_color_manual(values = c("Signifikan (p < 0.05)" = "#2c3e50", 
                                "Tidak Signifikan" = "#95a5a6")) +
  
  # F. Merapikan teks judul dan sumbu
  labs(
    title = "Determinan Kualitas Pengalaman Menarche pada Remaja",
    subtitle = "Forest Plot: Odds Ratio (OR) dan 95% Confidence Interval",
    x = "Odds Ratio (Skala Logaritmik)",
    y = "",
    color = "Status Inferensi:"
  ) +
  
  # G. Menggunakan tema minimalis agar fokus pada data, bukan pernak-pernik
  theme_minimal() +
  theme(
    plot.title = element_text(face = "bold", size = 15, hjust = 0.5),
    plot.subtitle = element_text(size = 12, hjust = 0.5, color = "gray30", margin = margin(b = 15)),
    axis.text.y = element_text(face = "bold", size = 12, color = "black"),
    axis.text.x = element_text(size = 11, color = "black"),
    axis.title.x = element_text(face = "bold", margin = margin(t = 15)),
    legend.position = "bottom",
    legend.title = element_text(face = "bold"),
    panel.grid.minor = element_blank(),
    panel.grid.major.y = element_blank() # Menghapus garis background horizontal
  )

# 5. Menampilkan Plot
print(plot_forest_female)
ggsave("forest_female.png", plot = plot_forest_female, width = 10, height = 8, dpi = 300)
```

```r
# Jika belum memiliki library caret, hilangkan tanda pagar di bawah ini
# install.packages("caret")
library(caret)

# 1. Mendapatkan probabilitas dari model
probabilitas <- predict(model, type = "response")

# 2. Mengubah probabilitas menjadi prediksi biner (Threshold 0.5)
prediksi_kelas <- ifelse(probabilitas > 0.5, 1, 0)

# 3. KUNCI PERBAIKAN: Ubah menjadi factor dengan level yang dikunci (0 dan 1)
prediksi_factor <- factor(prediksi_kelas, levels = c(0, 1))
aktual_factor <- factor(df$pengalaman_biner, levels = c(0, 1))

# 4. Membuat Confusion Matrix yang komprehensif menggunakan caret
# Positive = "1" karena kita ingin melihat keberhasilan memprediksi pengalaman "Baik"
hasil_evaluasi <- confusionMatrix(data = prediksi_factor, 
                                  reference = aktual_factor, 
                                  positive = "1")

# 5. Menampilkan hasil
print(hasil_evaluasi)
```

```r
cat("\n=========================================\n")
cat("   1. HASIL UJI MULTIKOLINIERITAS (VIF)  \n")
cat("=========================================\n")
# Mengambil nilai VIF dan memasukkannya ke Data Frame
nilai_vif <- vif(model)
tabel_vif <- data.frame(
  Variabel_Prediktor = names(nilai_vif),
  Nilai_VIF = round(nilai_vif, 3),
  Status = ifelse(nilai_vif < 5, "Aman (Tidak ada Multikolinieritas)", "Bahaya")
)
print(tabel_vif, row.names = FALSE)

cat("\n=========================================\n")
cat(" 2. UJI KELAYAKAN MODEL (HOSMER-LEMESHOW) \n")
cat("=========================================\n")
# Mengambil hasil HL Test dan menyusunnya menjadi tabel
hl_test <- hoslem.test(model$y, fitted(model), g=10)
tabel_hl <- data.frame(
  Chi_Square = round(hl_test$statistic, 4),
  Derajat_Bebas = hl_test$parameter,
  P_Value = round(hl_test$p.value, 4),
  Kesimpulan = ifelse(hl_test$p.value > 0.05, "Model Fit (Layak)", "Model Tidak Fit")
)
print(tabel_hl, row.names = FALSE)
```

```r
# 1. Menghitung nilai Cook's Distance untuk semua responden
jarak_cook <- cooks.distance(model)

# 2. Menghitung ukuran sampel (n) dan ambang batas yang ketat (4/n)
n <- nobs(model) # harusnya 958
ambang_batas <- 4 / n

print(paste("Ambang Batas Ketat (4/n):", round(ambang_batas, 4)))

# 3. Menampilkan 5 responden dengan nilai Cook's Distance tertinggi
top_5_observasi <- head(sort(jarak_cook, decreasing = TRUE), 5)
print("--- 5 Observasi Paling Berpengaruh ---")
print(top_5_observasi)

# 4. MEMBEDAH ANOMALI (Tugas Analisis Lanjutan)
# Mari kita lihat apa jawaban asli dari responden nomor 93, 130, dan 265
# (Ganti 'df' dengan nama dataframe kamu yang berisi data mentah)
data_anomali <- df[c(93, 130, 265), c("pengalaman_biner", "dukungan_biner", "sumber_biner")]
print("--- Profil Jawaban Responden Ekstrem ---")
print(data_anomali)
```

```r
# 1. Menyiapkan Data untuk Plot
n_sampel <- nobs(model)
ambang_batas <- 4 / n_sampel
jarak_cook <- cooks.distance(model)

# Membuat data frame khusus untuk visualisasi
df_cook <- data.frame(
  Nomor_Observasi = 1:n_sampel,
  Cooks_Distance = jarak_cook
)

# Menambahkan aturan: Siapa yang melebihi ambang batas?
df_cook <- df_cook %>%
  mutate(
    Status = ifelse(Cooks_Distance > ambang_batas, "Berpengaruh (Melewati Batas)", "Aman"),
    Label = ifelse(Cooks_Distance > ambang_batas, Nomor_Observasi, "") # Hanya beri label jika lewat batas
  )

# 2. Menggambar Plot dengan ggplot2
plot_cook <- ggplot(df_cook, aes(x = Nomor_Observasi, y = Cooks_Distance, color = Status)) +
  # Membuat garis vertikal (jarum)
  geom_segment(aes(x = Nomor_Observasi, xend = Nomor_Observasi, y = 0, yend = Cooks_Distance), alpha = 0.6) +
  # Menambahkan titik di ujung jarum
  geom_point(size = 1.5) +
  # Menambahkan garis ambang batas horizontal berwarna merah putus-putus
  geom_hline(yintercept = ambang_batas, linetype = "dashed", color = "red", linewidth = 1) +
  # Menambahkan label angka di atas titik yang bermasalah
  geom_text(aes(label = Label), vjust = -1, size = 3, color = "black", check_overlap = TRUE) +
  # Mengatur warna kustom
  scale_color_manual(values = c("Aman" = "steelblue", "Berpengaruh (Melewati Batas)" = "darkred")) +
  # Mengatur label dan judul
  labs(
    title = "Deteksi Observasi Berpengaruh (Cook's Distance)",
    subtitle = paste("Ambang Batas Ketat (4/n) =", round(ambang_batas, 4)),
    x = "Nomor Urut Responden",
    y = "Nilai Cook's Distance"
  ) +
  # Menggunakan tema minimalis agar terlihat profesional
  theme_minimal() +
  theme(
    legend.position = "bottom",
    plot.title = element_text(face = "bold", size = 14),
    axis.title = element_text(face = "bold")
  )

# Menampilkan Plot
print(plot_cook)
```

```r
ggsave("plot_cook_female.png", plot = plot_cook, width = 10, height = 8, dpi = 300)
```
